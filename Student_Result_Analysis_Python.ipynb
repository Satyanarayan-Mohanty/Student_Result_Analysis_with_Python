# Gender Distribution

```python
plt.figure(figsize = (5,5))
ax = sns.countplot(data = df, x = "gender")
ax.bar_label(ax.containers[0])
plt.title("Gender Distribution")
plt.show()
```

```python
# from the above chart we have analysed that :
# the number of males in the data is more than the number of females
```

# Parent Education impact on differnt children scores 

```python
gb =    df.groupby("parent_education").agg({"exam_score":'mean', "time_management_score":'mean'})
print(gb)
```

```python
sns.heatmap(gb)
plt.show
```

```python
sns.heatmap(gb, annot = True)
plt.title("Relationship between Parent's Education and Student's Score")
plt.show
```

```python
# From the above chart we have concluded that 
# the education of the parents have no/negligible impact on children score
```

```python
gb1 =    df.groupby("family_income").agg({"exam_score":'mean', "time_management_score":'mean'})
print(gb1)
```

```python
plt.figure(figsize = (4,4))
sns.heatmap(gb1, annot = True)
plt.title("Relationship between Family_Income and Student's Score")
plt.show
```

```python
# From the above chart we have concluded 
# there is no/negligible impact on student's score due to their family_income
```

```python
# For relationship findout purpose we use Heatmap .
```
